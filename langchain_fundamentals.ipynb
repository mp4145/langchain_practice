#### LangChain is an open-source framework that helps developers build applications using large language models (LLMs) like GPT or Claude

Langchain does the following: 
- Connects Data: It lets models read private files, live web pages, or databases (known as Retrieval-Augmented Generation or RAG).
- Creates Chains: It links multiple steps together, where the output of one task becomes the input for the next.
- Adds Memory: It helps a model remember past messages in a chat instead of starting fresh every time.
- Builds Agents: It allows the model to act as a reasoning engine that chooses which tools to use to solve a problem.

You can absolutely build all of these features without LangChain. LangChain does not provide any hidden "AI magic"; it is simply a wrapper and orchestration layer around standard code

```python
from langchain_ollama import ChatOllama

chat_model = ChatOllama(
    validate_model_on_init=True,
    model = "qwen3:4b",
    temperature = 0.6 #the amount of randomness the model uses to generate responses
)
```

```python
chat_model.invoke("Tell me a joke about Data Scientists")
```

#### Might be easy to use HumanMessage when sharing more than one sentence 

```python
from langchain_core.messages import HumanMessage

chat_model.invoke([
    HumanMessage("Tell me a joke about Data Scientists")
])
```

#### Prompt Template 

```python
from langchain_core.prompts import ChatPromptTemplate

joke_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a world class comedian."),
    ("human", "Tell me a joke about {topic}")
])
```

```python
joke_prompt.invoke({'topic': 'Data Scientists'})
```

```python
chain = joke_prompt | chat_model
```

```python
chain.invoke({"topic": "Data Scientists"})
```

#### Output Parser is responsible for parsing the output of the model into a more accessible format 

```python
from langchain_core.output_parsers import StrOutputParser

str_chain = chain | StrOutputParser()
```

```python
str_chain.invoke({"topic":"Data Scientists"})
```

#If you call .stream() on a Prompt Template, it will just yield a single chunk with the same output as .invoke()

```python
for chunk in str_chain.stream({'topic': 'Data Scientists'}): 
    print(chunk, end="|")   
```

#### Data Paradigm

```python

chat_model.invoke("What is the current date?")
```

```python
from datetime import date

date_prompt = ChatPromptTemplate.from_messages([
    ("system", 'You know that the current date is "{current_date}".'),
    ("human", "{question}")
])

date_chain = date_prompt | chat_model | StrOutputParser()

date_chain.invoke({
    "question": "What is the current date?",
    "current_date": date.today()
})
```

#### Importance of Context 

```python
(chat_model | StrOutputParser()).invoke(
    "What was the Old Ship Saloon's total revenue in Q1 2023?"
 )
```

```python
SOURCE = """
Old Ship Saloon 2023 quarterly revenue numbers:
Q1: $174782.38
Q2: $467372.38
Q3: $474773.38
Q4: $389289.23
"""

rag_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Use the following context when responding:\n\n{context}."),
    ("human", "{question}")
])

rag_chain = rag_prompt | chat_model | StrOutputParser()

rag_chain.invoke({
    "question": "What was the Old Ship Saloon's total revenue in Q1 2023?",
    "context": SOURCE
})
```

```python
from langchain.globals import set_debug 

set_debug(True)

date_prompt = ChatPromptTemplate.from_messages([
    ("system", 'You know that the current date is "{current_date}".'),
    ("human", "{question}")
])

date_chain = date_prompt | chat_model | StrOutputParser()

date_chain.invoke({
    "question": "What is the current date?",
    "current_date": date.today()
})
```

#### This is useful if you want to use intermediate steps in your application logic.

```python
# Turn off debug mode for clarity
set_debug(False)

stream = date_chain.astream_events({
    "question": "What is the current date?",
    "current_date": date.today()
}, version="v2")

async for event in stream:
    print(event)
    print("-----")
```
